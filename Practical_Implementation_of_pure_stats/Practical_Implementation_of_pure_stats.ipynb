#Measure of Central Tendency
1. Mean
2. Median
3. Mode

```python
ages = [23,24,32,45,12,43,67,45,32,56,32,120]
```

```python
import numpy as np
print(np.mean(ages))
print(np.median(ages))
```

```python
import statistics as st
print(st.mean(ages))
print(st.median(ages))
```

```python
st.mode(ages)
```

```python
import seaborn as sns
sns.boxplot(ages)
```

#5 Number Summry

```python
import numpy as np
q1, q3 = np.percentile(ages,[25,75])
print(q1,q3)
```

```python
#to check outlier[lower_fence and upper_fence]
IQR = q3-q1
lower_fence = q1-(1.5*IQR)
upper_fence = q3+(1.5*IQR)
print(lower_fence,upper_fence)
```

#Measure of Dispersion
1. Variance
2. Standard Deviation

```python
st.variance(ages)
```

```python
np.var(ages, axis =0)
```

```python
def variance(data,dof =0):
  n = len(data)
  mean = sum(data)/n
  deviations = [(x-mean)**2 for x in data]
  variance = sum(deviations)/(n-dof)
  return variance
```

```python
variance(ages, dof =1)
```

```python
st.pvariance(ages)
```

```python
import math
```

```python
math.sqrt(st.pvariance(ages))
```

#Check whether Distribution is Normal Distribution

```python
# If you want to check whether feature is guassian or normal distribution
#### Q-Q Plot
import matplotlib.pyplot as plt
import scipy.stats as stats
import pylab
def plot_data(sample):
  plt.figure(figsize =(12,4))
  plt.subplot(1,2,1)
  sns.histplot(sample, kde = True)
  plt.subplot(1,2,2)
  stats.probplot(sample, dist = "norm", plot = pylab)
  plt.show()
```

```python
#create a normal distributed data
s = np.random.normal(0.5,0.2,1000)
plot_data(s)
```

```python
plot_data(df['sepal_width'])
```
